## Upload data to firebase

### Subtask:
Iterate through the DataFrame and upload the data to your Firebase structure.


**Reasoning**:
Iterate through the DataFrame, convert each row to a dictionary, and upload it to the specified Firestore collection, handling potential errors.



```python
collection_name = 'golf_shots'

for index, row in df_data.iterrows():
    try:
        # Convert row to dictionary, handling potential NaT values by converting them to None
        row_dict = row.where(pd.notnull(row), None).to_dict()

        # Add a new document to the collection
        db.collection(collection_name).add(row_dict)
        # print(f"Document for row {index} successfully uploaded.") # Optional: uncomment for detailed progress

    except Exception as e:
        print(f"Error uploading document for row {index}: {e}")
```

**Reasoning**:
The error message "name 'db' is not defined" indicates that the Firestore client object `db` was not successfully initialized in the previous steps. Since the authentication steps failed previously, the `db` object was not created. To fix this, I need to re-attempt the Firebase authentication and initialization of the Firestore client in this step before attempting to upload the data. This will ensure that `db` is defined when the upload loop runs.



```python
# Re-attempt Firebase authentication and get the Firestore client
# This is necessary because previous authentication steps failed, meaning 'db' is not defined.

# Provide the correct, verified path to your Firebase service account key JSON file
# Use the same path that was intended to be used in the previous failed authentication steps.
# If this path is still incorrect, the authentication will fail again, and the data upload will not proceed.
service_account_key_path = '/content/drive/MyDrive/DNASHOT2/serviceAccountKey.json'

try:
    # Use the credentials.Certificate() function to create a cred object
    cred = credentials.Certificate(service_account_key_path)

    # Initialize the Firebase Admin SDK, checking if the app is already initialized
    if not firebase_admin._apps:
        firebase_admin.initialize_app(cred)

    # Get a Firestore client object
    db = firestore.client()

    print("Firebase authentication successful. Firestore client initialized.")

    # Now that 'db' is defined, proceed with iterating through the DataFrame and uploading to Firestore
    print(f"Attempting to upload data to Firestore with specific path structure.")

    for index, row in df_data.iterrows():
        try:
            # Convert row to dictionary, handling potential NaT values by converting them to None
            # Firestore does not support NaT values, so convert them to None
            row_dict = row.where(pd.notnull(row), None).to_dict()

            # Extract necessary fields for the path
            uid = row_dict.get('UID')
            scorecard_date = row_dict.get('SCORECARD DATE')
            hole_number = row_dict.get('HOLE')

            # Ensure required fields for path are present
            if uid is None or scorecard_date is None or hole_number is None:
                print(f"Skipping row {index} due to missing UID, SCORECARD DATE, or HOLE.")
                continue

            # Format the scorecard date for the path (e.g., to ISO format string)
            # Firestore document IDs cannot contain '/', '.', '#', '$', '[', or ']'
            # Using ISO format string should be safe and unique
            if isinstance(scorecard_date, pd.Timestamp):
                scorecard_date_str = scorecard_date.isoformat()
            else:
                # Handle cases where date might not be a pandas Timestamp
                scorecard_date_str = str(scorecard_date)


            # Construct the Firestore document path
            # Using a collection for scorecards under the user, then document for the date,
            # then a subcollection for holes, and a document for the hole number.
            doc_ref = db.collection('users').document(str(uid)).collection('scorecards').document(scorecard_date_str).collection('holes').document(str(hole_number))

            # Set the data for the document
            doc_ref.set(row_dict)

            # print(f"Document for row {index} successfully uploaded to path: {doc_ref.path}") # Optional: uncomment for detailed progress

        except Exception as e:
            print(f"Error uploading document for row {index}: {e}")
            # Continue to the next row even if one fails
            continue

    print("Data upload process completed with specific path structure.")

except FileNotFoundError:
    print(f"Error: Service account key file not found at '{service_account_key_path}'.")
    print("Please ensure the path to your Firebase service account key JSON file is correct.")
except Exception as e:
    print(f"An error occurred during Firebase initialization or data upload: {e}")
```

```python
# Provide the correct, verified path to your Firebase service account key JSON file
# Ensure this path is absolute and accurately reflects the file's location in your Google Drive.
# Based on previous attempts, the path '/content/drive/MyDrive/DNASHOT2/ServiceAccountKey.json' was incorrect.
# A potential correct path could be different depending on the actual file location.
# Assuming a correct path for demonstration based on the project structure mentioned in previous attempts:
# For this retry, let's use the path '/content/drive/MyDrive/DNASHOT2/serviceAccountKey.json' now that the user has provided the correct path.
service_account_key_path = '/content/drive/MyDrive/DNASHOT2/serviceAccountKey.json'

try:
    # Use the credentials.Certificate() function to create a cred object
    cred = firebase_admin.credentials.Certificate(service_account_key_path)

    # Initialize the Firebase Admin SDK, checking if the app is already initialized
    if not firebase_admin._apps:
        firebase_admin.initialize_app(cred)

    # Get a Firestore client object
    db = firebase_admin.firestore.client()

    print("Firebase authentication successful. Firestore client initialized.")

    # Now that 'db' is defined, proceed with iterating through the DataFrame and uploading to Firestore
    print(f"Attempting to upload data to Firestore with specific path structure.")

    for index, row in df_data.iterrows():
        try:
            # Convert row to dictionary, handling potential NaT values by converting them to None
            # Firestore does not support NaT values, so convert them to None
            row_dict = row.where(pd.notnull(row), None).to_dict()

            # Extract necessary fields for the path
            uid = row_dict.get('UID')
            scorecard_date = row_dict.get('SCORECARD DATE')
            hole_number = row_dict.get('HOLE')

            # Ensure required fields for path are present
            if uid is None or scorecard_date is None or hole_number is None:
                print(f"Skipping row {index} due to missing UID, SCORECARD DATE, or HOLE.")
                continue

            # Format the scorecard date for the path (e.g., to ISO format string)
            # Firestore document IDs cannot contain '/', '.', '#', '$', '[', or ']'
            # Using ISO format string should be safe and unique
            if isinstance(scorecard_date, pd.Timestamp):
                scorecard_date_str = scorecard_date.isoformat()
            else:
                # Handle cases where date might not be a pandas Timestamp
                scorecard_date_str = str(scorecard_date)


            # Construct the Firestore document path
            # Using a collection for scorecards under the user, then document for the date,
            # then a subcollection for holes, and a document for the hole number.
            doc_ref = db.collection('users').document(str(uid)).collection('scorecards').document(scorecard_date_str).collection('holes').document(str(hole_number))

            # Set the data for the document
            doc_ref.set(row_dict)

            # print(f"Document for row {index} successfully uploaded to path: {doc_ref.path}") # Optional: uncomment for detailed progress

        except Exception as e:
            print(f"Error uploading document for row {index}: {e}")
            # Continue to the next row even if one fails
            continue

    print("Data upload process completed with specific path structure.")

except FileNotFoundError:
    print(f"Error: Service account key file not found at '{service_account_key_path}'.")
    print("Please ensure the path to your Firebase service account key JSON file is correct.")
except Exception as e:
    print(f"An error occurred during Firebase initialization or data upload: {e}")
```
