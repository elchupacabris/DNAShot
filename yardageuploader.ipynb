## Prepare data for structured upload

### Subtask:
Restructure the `df_yardage` DataFrame to group data by course and tee, and then prepare it for upload as nested documents with subcollections for holes.


**Reasoning**:
Group the DataFrame by 'CC' and 'TEE' and then iterate through the groups to create a nested structure for uploading to Firestore, with each course/tee combination as a document and its holes in a subcollection.



```python
courses_data = {}

grouped = df_yardage.groupby(['CC', 'TEE'])

for (course_name, tee_name), group in grouped:
    course_key = f"{course_name}_{tee_name}"
    courses_data[course_key] = {
        'CC': course_name,
        'TEE': tee_name,
        'holes': []
    }
    for index, row in group.iterrows():
        hole_data = {
            'HOLE': row['HOLE'],
            'PAR': row['PAR'],
            'YDS': row['YDS'],
            'VENTAJA': row['VENTAJA']
        }
        courses_data[course_key]['holes'].append(hole_data)

print(f"Prepared data structure for {len(courses_data)} courses/tees.")
```

**Reasoning**:
The data has been restructured into a nested dictionary format, where each key represents a course and tee combination and contains a list of hole data. The next logical step is to upload this structured data to the 'courses' collection in Firestore. For each course/tee entry, a document will be created with the course and tee names as fields and a 'holes' subcollection containing the hole data.



```python
collection_ref = db.collection('courses')

for course_key, course_data in courses_data.items():
    course_doc_ref = collection_ref.document(course_key)
    # Create the course document with course and tee info
    course_doc_ref.set({
        'CC': course_data['CC'],
        'TEE': course_data['TEE']
    })

    # Create a subcollection 'holes' for each course document
    holes_collection_ref = course_doc_ref.collection('holes')
    for hole in course_data['holes']:
        # Add each hole as a document in the 'holes' subcollection
        holes_collection_ref.add(hole)

print("Data upload to Firestore initiated.")
```

Restructure the DataFrame to group data by user (UID), and then prepare it for upload with nested 'CC' subcollections and hole data within those.


**Reasoning**:
Group the DataFrame by 'UID' and then iterate through the groups to create a nested structure for uploading to Firestore, with each user as a document and their course and hole data in nested structures within.



```python
users_data_structured = {}

grouped_by_user = df_all_data.groupby('UID')

for uid, user_group in grouped_by_user:
    users_data_structured[uid] = {
        'courses': {}
    }
    # Group by course and tee within each user's data
    grouped_by_course_tee = user_group.groupby(['CC', 'TEE'])
    for (cc, tee), course_tee_group in grouped_by_course_tee:
        course_tee_key = f"{cc}_{tee}"
        users_data_structured[uid]['courses'][course_tee_key] = {
            'CC': cc,
            'TEE': tee,
            'holes': {}
        }
        # Iterate through holes within each course/tee
        for index, hole_row in course_tee_group.iterrows():
            hole_number = str(hole_row['HOLE'])
            users_data_structured[uid]['courses'][course_tee_key]['holes'][hole_number] = {
                'HOLE': hole_row['HOLE'],
                'PAR': hole_row['PAR'],
                'YDS': hole_row['YDS'],
                'SCORE': hole_row['SCORE'],
                'PUTT': hole_row['PUTT'],
                'DD': hole_row['DD'],
                'DDTYPE': hole_row['DDTYPE'],
                'PINDIST': hole_row['PINDIST'],
                '1STPU': hole_row['1STPU'],
                'ESC': hole_row['ESC']
            }

print(f"Prepared structured data for {len(users_data_structured)} users.")
```
